```python
import numpy as np
from imps import iMPS
```

```python
def pcoin(p):
    """
    transition matrix for pcoin

    -------------------
    Parameters
    -------------------
    p:float
    -------------------
    
    """
    T0 = np.array([
        [1.-p,p],
        [0.,0.]
    ])
    T1 = np.array([
        [0.,0.],
        [p,1.-p]
    ])
    return np.array([T0,T1])
```

```python
p = 0.1
imps_pcoin = iMPS.from_tmatrix(pcoin(p))
U = imps_pcoin.to_unitary()
```
